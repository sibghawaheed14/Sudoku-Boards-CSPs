```python
# CREATE INPUT FILES
with open("easy.txt", "w") as f:
    f.write("""004030050
609400000
005100489
000060930
300807002
026040000
453009600
000004705
090050200""")

with open("medium.txt", "w") as f:
    f.write("""000030040
109700000
000851070
002607830
906010207
031502900
010369000
000005703
090070000""")

with open("hard.txt", "w") as f:
    f.write("""102040007
000080000
009500304
000607900
540000026
006405000
708003400
000010000
200060509""")

with open("veryhard.txt", "w") as f:
    f.write("""001007000
600400300
000030064
380076000
000000036
270015000
000020051
700100200
008009000""")


# GLOBAL VARIABLES FOR COUNTERS
backtrack_calls = 0
failures = 0


# READING BOARD
def read_board(filename):
    board = []
    with open(filename, 'r') as f:
        for line in f:
            board.append([int(x) for x in line if x != '\n'])

    return board


# PRINTING BOARD
def print_board(board):
    for row in board:
        print(" ".join(str(x) for x in row))


# FINDINNG EMPTY CELL
def find_empty(board):
    for i in range(9):
        for j in range(9):
            if board[i][j] == 0:
                return i, j
    return None


# CHECKING VALIDITY
def is_valid(board, row, col, num):

    # Row check
    if num in board[row]:
        return False

    for i in range(9):
        if board[i][col] == num:
            return False

    start_row = row - row % 3
    start_col = col - col % 3

    for i in range(3):
        for j in range(3):
            if board[start_row + i][start_col + j] == num:
                return False

    return True


# AC-3
def ac3(board):
    changed = True

    while changed:
        changed = False

        for i in range(9):
            for j in range(9):

                if board[i][j] == 0:
                    possible = []

                    for num in range(1, 10):
                        if is_valid(board, i, j, num):
                            possible.append(num)

                    if len(possible) == 1:
                        board[i][j] = possible[0]
                        changed = True

    return board


# FORWARD CHECKING
def forward_check(board, row, col):

    for i in range(9):

        if board[row][i] == 0:
            if not any(is_valid(board, row, i, num) for num in range(1, 10)):
                return False

        if board[i][col] == 0:
            if not any(is_valid(board, i, col, num) for num in range(1, 10)):
                return False

    start_row = row - row % 3
    start_col = col - col % 3

    for i in range(3):
        for j in range(3):
            r = start_row + i
            c = start_col + j

            if board[r][c] == 0:
                if not any(is_valid(board, r, c, num) for num in range(1, 10)):
                    return False

    return True


# BACKTRACKING
def backtrack(board):
    global backtrack_calls, failures

    backtrack_calls += 1

    empty = find_empty(board)
    if not empty:
        return True

    row, col = empty

    for num in range(1, 10):

        if is_valid(board, row, col, num):
            board[row][col] = num

            temp_board = [r[:] for r in board]

            ac3(temp_board)

            if forward_check(temp_board, row, col):
                if backtrack(temp_board):

                    for i in range(9):
                        board[i] = temp_board[i][:]

                    return True

            board[row][col] = 0

    failures += 1
    return False



files_list = ["easy.txt", "medium.txt", "hard.txt", "veryhard.txt"]

for file in files_list:

    print("\nSolving:", file)

    backtrack_calls = 0
    failures = 0

    board = read_board(file)

    if backtrack(board):
        print("\nFinal Board:")
        print_board(board)
    else:
        print("No solution found")

    print("\nBacktrack Calls:", backtrack_calls)
    print("Failures:", failures)

    
```
